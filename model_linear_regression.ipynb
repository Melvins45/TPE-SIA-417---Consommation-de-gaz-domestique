<h2> Nettoyage des données </h2>

```python
df_tpe.columns = ["timestamp", "nb_personnes", "marque", "equipement", "nb_foyers_moyenne", "heures_utilisation", "temps_consommation", "ville"]

df_tpe = df_tpe.rename(columns={"nb_personnes" : "plage_nb_personnes", "heures_utilisation" : "plage_heures_utilisation"})

print(df_tpe.describe())
```

```python
df_tpe["ville"] = df_tpe["ville"].fillna(df_tpe["ville"].mode()[0])
df_tpe["plage_nb_personnes"] = df_tpe["plage_nb_personnes"].fillna(df_tpe["plage_nb_personnes"].mode()[0])

print(df_tpe.head())
print("Pourcentage de valeurs manquantes : ")
print(df_tpe.isna().mean() * 100)
```

```python
assert df_tpe.isna().sum().sum() == 0
print("OK, aucune valeur manquante trouvée.")
```

```python
print(df_tpe["ville"].unique())
```

```python
import re

villes = ["Douala", "Yaoundé", "Nkolbong", "Mbouda", "Kribi", "Garoua", "Ouagadougou", "Bafoussam", "Ndogbong", "Bonaberi"]
pattern = r"\b(" + "|".join(map(re.escape, villes)) + r")\b"
villes_map = {
    "bonaberi" : "Douala",
    "ndogbong" : "Douala",
    "nkolbong" : "Douala",
    "ouagadougou" : "Ouagadougou",
    "yaoundé" : "Yaoundé",
    "douala" : "Douala",
    "kribi" : "Kribi",
    "bafoussam" : "Bafoussam",
    "garoua" : "Garoua",
    "mbouda" : "Mbouda"
}

df_tpe["ville"] = df_tpe["ville"].str.extract(pattern, flags=re.IGNORECASE, expand=False).str.lower().map(villes_map)

print("Affichons les résultats : ")
print(df_tpe["ville"].head())
print("Valeurs uniques : ", df_tpe["ville"].unique())
print("Valeurs nulles : ", df_tpe["ville"].isna().sum())
```

```python
columns_not_number = df_tpe.select_dtypes(exclude="number").columns.tolist()

print(columns_not_number)

message = "Les colonnes non numériques avec leurs valeurs distinctes sont :\n"

for column in columns_not_number[1:] :
    message += f"\t- {column} : {df_tpe[column].nunique()} valeurs uniques qui sont {df_tpe[column].unique()}\n"
    
print(message)
```

```python
marques_map = {
    "SCTM" : "SCTM",
    "Camgaz" : "Camgaz",
    "Tradex" : "Tradex",
    "Bocom" : "Bocom",
    "Total" : "Total",
    "Ola " : "Ola",
    "Tradex " : "Tradex",
    "Cameroun Gaz" : "Camgaz",
    "Mobile (OLA)" : "Ola",
    "Ola" : "Ola",
    "Oryx" : "Oryx",
    "Bouteille de 6kg (petit format)" : "PetitFormat",
    "Cap gaz" : "Camgaz",
    "Total " : "Total",
    "BOCOM et SCTM" : "Bocom",
    "Bocom " : "Bocom",
    "Cam gaz" : "Camgaz",
    "Olybia" : "Olybia",
    "OLA" : "Ola",
    "MRS 15kg" : "MRS"
}

df_tpe["marque"] = df_tpe["marque"].map(marques_map)

print("Affichons les résultats : ")
print(df_tpe["marque"].head())
print("Valeurs distinctes : ", df_tpe["marque"].unique())
print("Valeurs manquantes : ", df_tpe["marque"].isna().sum())
```

```python
equipement_map = {
    "Ecogaz et cuisine " : "Ecogaz"
}

df_tpe["equipement"] = df_tpe["equipement"].replace(equipement_map)

print("Affichons les résultats : ")
print(df_tpe["equipement"].head())
print("Valeurs distinctes : ", df_tpe["equipement"].unique())
print("Valeurs manquantes : ", df_tpe["equipement"].isna().sum())
```

```python
nb_foyers_moyenne_map = {
    "3 foyers ou plus" : "3",
    "2 foyers" : "2",
    "1 seul foyer" : "1"
}

df_tpe["nb_foyers_moyenne"] = (df_tpe["nb_foyers_moyenne"].replace(nb_foyers_moyenne_map)).astype(int)

print("Affichons les résultats : ")
print(df_tpe["nb_foyers_moyenne"].head())
print("Valeurs distinctes : ", df_tpe["nb_foyers_moyenne"].unique())
print("Valeurs manquantes : ", df_tpe["nb_foyers_moyenne"].isna().sum())
```

```python
plage_heures_utilisation_map = {
    "Plus de 4 heures" : "4.5",
    "1 à 2 heures par jour" : "1.5",
    "2h01min  à 4 heures par jour" : "3.0",
    "Moins d’une heure par jour" : "0.5"
}

df_tpe["plage_heures_utilisation"] = (df_tpe["plage_heures_utilisation"].replace(plage_heures_utilisation_map)).astype(float)

print("Affichons les résultats : ")
print(df_tpe["plage_heures_utilisation"].head())
print("Valeurs distinctes : ", df_tpe["plage_heures_utilisation"].unique())
print("Valeurs manquantes : ", df_tpe["plage_heures_utilisation"].isna().sum())
```

```python
temps_consommation_map = {
    "1 mois environ" : "1.0",
    "1 mois et demi à 2 mois" : "1.5",
    "Plus de 2 mois" : "2.5",
    "moins de 2 semaines" : "0.5"
}

df_tpe["temps_consommation"] = (df_tpe["temps_consommation"].replace(temps_consommation_map)).astype(float)

print("Affichons les résultats : ")
print(df_tpe["temps_consommation"].head())
print("Valeurs distinctes : ", df_tpe["temps_consommation"].unique())
print("Valeurs manquantes : ", df_tpe["temps_consommation"].isna().sum())
```

```python
plage_nb_personnes_map = {
    "6 personnes et plus" : "6.5",
    "4 à 5 personnes" : "4.5",
    "2 à 3 personnes" : "2.5",
    "1 personne" : "1.0"
}

df_tpe["plage_nb_personnes"] = (df_tpe["plage_nb_personnes"].replace(plage_nb_personnes_map)).astype(float)

print("Affichons les résultats : ")
print(df_tpe["plage_nb_personnes"].head())
print("Valeurs distinctes : ", df_tpe["plage_nb_personnes"].unique())
print("Valeurs manquantes : ", df_tpe["plage_nb_personnes"].isna().sum())
```

```python
columns_not_number = df_tpe.select_dtypes(exclude="number").columns.tolist()

print(columns_not_number)

message = "Les colonnes  avec leurs valeurs distinctes sont :\n"

for column in df_tpe.columns.tolist()[1:] :
    message += f"\t- {column} : {df_tpe[column].nunique()} valeurs uniques qui sont {df_tpe[column].unique()}\n"
    
print(message)
```
